# Semantic Resonance Budget (SRB) – Quickstart

This notebook demonstrates a minimal end-to-end SRB analysis on a synthetic
token sequence, using the `srb` library.

```python
import math

from srb import SRBConfig, TokenStep, analyze_sequence

# Build a tiny synthetic sequence along a 1D line in embedding space.
steps = []
probs = [0.7, 0.3]
logprob = math.log(probs[0])

for i in range(4):
    steps.append(
        TokenStep(
            index=i,
            token=f"tok_{i}",
            logprob=logprob,
            probs=probs,
            embedding=[float(i), 0.0],
        )
    )

config = SRBConfig()
result = analyze_sequence(steps, config=config)

print("Per-step metrics:")
for s in result.steps:
    print(
        f"idx={s.index} | v={s.semantic_velocity!r} | "
        f"H={s.semantic_entropy!r} | S={s.surprisal!r} | "
        f"dS={s.surprisal_gradient!r} | NSM={s.nsm_divergence!r} | "
        f"a={s.semantic_acceleration!r}"
    )

print("\nAggregates:")
for name, value in result.aggregates.items():
    print(f"{name}: {value}")
```

```python
from srb.viz import plot_metric_series

_ = plot_metric_series(result, "semantic_velocity", title="Semantic Velocity")
```
